```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

harga * 0.9 pada NumPy array melakukan perkalian elementwise (tiap elemen dikali 0.9) — ini disebut vectorization, karena NumPy dirancang untuk operasi numerik.

Pada list Python biasa, operator * berarti "mengulang isi list", bukan mengalikan tiap elemen, dan hanya berlaku untuk integer. Karena 0.9 bukan integer, muncul TypeError.

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom
```

menu (baris ke-5) dan terjual (baris ke-3). perhitungan statistik (rata-rata, total) jadi bias, analisis per menu tidak akurat karena tidak tahu produknya, dan bisa menyebabkan error saat diproses lebih lanjut (misal groupby atau machine learning). Makanya data kosong perlu dibersihkan dulu sebelum dianalisis.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Total baris: 110. Kolom yang non-null-nya lebih sedikit dari 110:

menu → 107 non-null (ada 3 data kosong)
terjual → 104 non-null (ada 6 data kosong)

Kolom tanggal, kategori, dan harga semuanya lengkap (110 non-null).

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

terjual diisi 0 (fillna) karena kolom numerik, wajar dianggap "belum ada transaksi", dan barisnya masih berguna untuk analisis lain.

Baris menu kosong dihapus (dropna) karena menu adalah identitas/kategori — tidak ada nilai netral pengganti nama menu, dan tanpa itu baris jadi tidak berguna untuk analisis per-menu.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

dari 107 → 102 baris (5 baris duplikat dihapus).
kalau tipe data salah (misal angka terbaca sebagai teks), perhitungan seperti sum()/mean() bisa error atau hasilnya salah tanpa disadari. Tipe data yang benar memastikan analisis numerik akurat dan efisien.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
